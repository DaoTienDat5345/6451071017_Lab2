```python
# ========================================================
# BÀI 5 (TRANG 26, 27) - CÂU A
# Đề bài: Nhập dữ liệu đầu vào và tiến hành tiền xử lý xóa các dòng dữ liệu trống,
# sửa tên các cột dữ liệu lại như sau để dễ theo dõi:
# 'age', 'sex', 'chest_pain_type', 'resting_blood_pressure', 'cholesterol',
# 'fasting_blood_sugar', 'rest_ecg', 'max_heart_rate_achieved',
# 'exercise_induced_angina', 'st_depression', 'st_slope', 'num_major_vessels',
# 'thalassemia', 'target'.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


# 2. Đọc file heart.csv
heart_path = find_file("heart.csv")
data = pd.read_csv(heart_path)
print(f"Số dòng và số cột ban đầu: {data.shape}")

# 3. Tiền xử lý: xóa dòng dữ liệu trống (nếu có)
data = data.dropna()
print(f"Số dòng sau khi xóa dữ liệu trống: {len(data)}")

# 4. Đổi tên các cột theo đúng yêu cầu đề bài
new_columns = [
    'age', 'sex', 'chest_pain_type', 'resting_blood_pressure', 'cholesterol',
    'fasting_blood_sugar', 'rest_ecg', 'max_heart_rate_achieved',
    'exercise_induced_angina', 'st_depression', 'st_slope', 'num_major_vessels',
    'thalassemia', 'target'
]
data.columns = new_columns

# 5. Hiển thị thông tin khối dữ liệu sau tiền xử lý
print("-" * 60)
print("Thông tin kiểu dữ liệu các cột mới:")
data.info()
print("-" * 60)
print("5 dòng đầu tiên:")
display(data.head())
```
