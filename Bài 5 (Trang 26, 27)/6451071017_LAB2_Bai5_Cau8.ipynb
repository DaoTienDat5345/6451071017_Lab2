```python
# ========================================================
# BÀI 5 (TRANG 26, 27) - CÂU 8
# Đề bài: Lặp lại yêu cầu 7 đối với cây CART (criterion='gini') trên dữ liệu Heart Disease.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn import tree, metrics
import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


data = pd.read_csv(find_file("heart.csv")).dropna()
data.columns = [
    'age', 'sex', 'chest_pain_type', 'resting_blood_pressure', 'cholesterol',
    'fasting_blood_sugar', 'rest_ecg', 'max_heart_rate_achieved',
    'exercise_induced_angina', 'st_depression', 'st_slope', 'num_major_vessels',
    'thalassemia', 'target'
]
features = data.drop('target', axis=1)
labels = data['target']
cat_cols = ['chest_pain_type', 'rest_ecg', 'st_slope', 'thalassemia']
features_onehot = pd.get_dummies(features, columns=cat_cols)

X_train, X_test, y_train, y_test = train_test_split(features_onehot, labels, test_size=0.3, random_state=42)

# 2. Xây dựng cây CART (criterion='gini')
clf_cart = tree.DecisionTreeClassifier(criterion='gini', random_state=42)
clf_cart.fit(X_train, y_train)

# 3. Dự đoán trên tập kiểm thử
cart_pred = clf_cart.predict(X_test)

# 4. Đánh giá độ chính xác
cart_score = metrics.accuracy_score(y_test, cart_pred)
print("Accuracy (CART):", cart_score)
print("Report:\n", metrics.classification_report(y_test, cart_pred))

# 5. Ma trận nhầm lẫn Heatmap
cart_cm = metrics.confusion_matrix(y_test, cart_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cart_cm, annot=True, fmt="d", cmap='Blues_r', square=True,
            xticklabels=['No Disease', 'Disease'], yticklabels=['No Disease', 'Disease'])
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
plt.title(f"CART Decision Tree (Heart Disease) - Accuracy: {cart_score:.4f}", size=14)
plt.show()

# 6. Biểu diễn cây CART
fig, ax = plt.subplots(figsize=(22, 10))
tree.plot_tree(clf_cart, max_depth=3, feature_names=list(features_onehot.columns), class_names=['No Disease', 'Disease'], filled=True, fontsize=9)
plt.title("Biểu diễn cấu trúc cây CART (max_depth=3) - Heart Disease", fontsize=15)
plt.savefig('heart_decision_tree_CART.png', dpi=100)
plt.show()
```
