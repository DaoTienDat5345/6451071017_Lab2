```python
# ========================================================
# BÀI 5 (TRANG 26, 27) - CÂU 5
# Đề bài: Chuyển đổi các cột không phải dạng số hoặc các thuộc tính định danh
# về dạng one-hot vector để phù hợp với đầu vào của thuật toán.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


data = pd.read_csv(find_file("heart.csv")).dropna()
data.columns = [
    'age', 'sex', 'chest_pain_type', 'resting_blood_pressure', 'cholesterol',
    'fasting_blood_sugar', 'rest_ecg', 'max_heart_rate_achieved',
    'exercise_induced_angina', 'st_depression', 'st_slope', 'num_major_vessels',
    'thalassemia', 'target'
]

features = data.drop('target', axis=1)

# 2. Các thuộc tính rời rạc / phân loại trong Heart Disease:
# chest_pain_type (0-3), rest_ecg (0-2), st_slope (0-2), thalassemia (0-3)
cat_cols = ['chest_pain_type', 'rest_ecg', 'st_slope', 'thalassemia']
print("Các cột rời rạc phân loại được chuyển đổi one-hot:")
print(cat_cols)
print("-" * 60)

# 3. Mã hóa One-hot vector
features_onehot = pd.get_dummies(features, columns=cat_cols, drop_first=False)

print(f"Kích thước features ban đầu: {features.shape}")
print(f"Kích thước features sau khi mã hóa one-hot: {features_onehot.shape}")
print("-" * 60)
print("5 dòng đầu sau khi chuyển đổi dạng one-hot:")
display(features_onehot.head())
```
