```python
# ========================================================
# BÀI 4 (TRANG 26) - CÂU 5
# Đề bài: Chuyển đổi các cột không phải dạng số về dạng one-hot vector
# để phù hợp với đầu vào của thư viện (áp dụng trên dữ liệu Wine Quality).
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


wine_path = find_file("winequality-red.csv")
data = pd.read_csv(wine_path)
features = data.drop('quality', axis=1)

# 2. Kiểm tra kiểu dữ liệu các cột trong features
print("Kiểu dữ liệu của các cột trong features:")
print(features.dtypes)
print("-" * 60)

# Xác định các thuộc tính không phải dạng số
categorical_cols = features.select_dtypes(exclude=['int64', 'float64']).columns
print(f"Số lượng cột không phải dạng số: {len(categorical_cols)}")
if len(categorical_cols) == 0:
    print("=> Tất cả các thuộc tính của dữ liệu Red Wine Quality đều là dữ liệu số liên tục (float64).")
    print("=> Do đó, không cần chuyển đổi dạng one-hot vector, giữ nguyên tập đặc trưng features.")
    features_onehot = features.copy()
else:
    features_onehot = pd.get_dummies(features, columns=categorical_cols)

print(f"\nKích thước tập đặc trưng sẵn sàng huấn luyện: {features_onehot.shape}")
display(features_onehot.head())
```
