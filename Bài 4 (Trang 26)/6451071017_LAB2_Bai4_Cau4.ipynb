```python
# ========================================================
# BÀI 4 (TRANG 26) - CÂU 4
# Đề bài: Tách các cột dữ liệu thành hai phần: một phần chứa các thuộc tính
# bình thường (features), một phần chứa riêng thuộc tính quyết định 'quality' (labels).
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


wine_path = find_file("winequality-red.csv")
data = pd.read_csv(wine_path)

# 2. Tách dữ liệu thành features và labels
features = data.drop('quality', axis=1)
labels = data['quality']

print(f"Kích thước tập thuộc tính features: {features.shape}")
print(f"Kích thước thuộc tính quyết định labels (quality): {labels.shape}")
print("-" * 60)
print("Danh sách các cột đặc trưng trong features:")
print(list(features.columns))
print("-" * 60)
print("Phân phối nhãn chất lượng rượu (quality):")
print(labels.value_counts().sort_index())
print("-" * 60)
display(features.head())
```
