```python
# ========================================================
# BÀI 4 (TRANG 26) - CÂU 9
# Đề bài: Xây dựng mô hình phân lớp bằng thuật toán Naïve Bayes
# trên dữ liệu Red Wine Quality và kiểm tra kết quả đạt được.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn import metrics
import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


data = pd.read_csv(find_file("winequality-red.csv"))
X = data.drop('quality', axis=1)
y = data['quality']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 2. Xây dựng mô hình Gaussian Naïve Bayes
gnb = GaussianNB()
bayes_pred = gnb.fit(X_train, y_train).predict(X_test)

# 3. Đánh giá mô hình
bayes_score = metrics.accuracy_score(y_test, bayes_pred)
print("Accuracy (Naïve Bayes):", bayes_score)
print("Report:\n", metrics.classification_report(y_test, bayes_pred, zero_division=0))

# 4. Biểu diễn ma trận nhầm lẫn
bayes_cm = metrics.confusion_matrix(y_test, bayes_pred)
plt.figure(figsize=(9, 7))
sns.heatmap(bayes_cm, annot=True, fmt="d", cmap='Blues_r', square=True,
            xticklabels=gnb.classes_, yticklabels=gnb.classes_)
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
plt.title(f"Naïve Bayes (Wine Quality) - Accuracy: {bayes_score:.4f}", size=14)
plt.show()
```
