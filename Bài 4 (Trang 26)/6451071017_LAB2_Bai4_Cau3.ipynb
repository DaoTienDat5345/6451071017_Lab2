```python
# ========================================================
# BÀI 4 (TRANG 26) - CÂU 3
# Đề bài: Khảo sát độ tương đồng giữa các cột với nhau bằng công thức Pearson
# trên tập dữ liệu Red Wine Quality (winequality-red.csv).
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


# 2. Đọc dữ liệu Red Wine Quality
wine_path = find_file("winequality-red.csv")
data = pd.read_csv(wine_path)
print(f"Kích thước tập dữ liệu: {data.shape}")

# 3. Khảo sát độ tương đồng Pearson
corr_matrix = data.corr(method='pearson')
print("Ma trận tương quan Pearson giữa các thuộc tính:")
display(corr_matrix)

# 4. Trực quan hóa bằng Heatmap
plt.figure(figsize=(14, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title("Heatmap tương quan Pearson dữ liệu Red Wine Quality", fontsize=15)
plt.show()

# 5. Nhận xét
print("=" * 60)
print("NHẬN XÉT:")
print("- Tương quan dương cao nhất: fixed acidity và citric acid (0.67), fixed acidity và density (0.67).")
print("- Tương quan âm cao nhất: fixed acidity và pH (-0.68), citric acid và pH (-0.54).")
print("- Thuộc tính có tương quan dương mạnh nhất với chất lượng rượu (quality) là alcohol (0.48).")
print("- Không có thuộc tính nào vượt ngưỡng đa cộng tuyến nghiêm trọng (> 0.85), giữ lại toàn bộ các cột.")
```
