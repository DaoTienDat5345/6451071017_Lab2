```python
# ========================================================
# BÀI 4 (TRANG 26) - CÂU 10
# Đề bài: So sánh kết quả của các mô hình trên (ID3, CART, Naïve Bayes)
# trên tập dữ liệu Red Wine Quality.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

%matplotlib inline
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn import tree, metrics
from sklearn.naive_bayes import GaussianNB
import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


data = pd.read_csv(find_file("winequality-red.csv"))
X = data.drop('quality', axis=1)
y = data['quality']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# 1. Cây ID3
clf_id3 = tree.DecisionTreeClassifier(criterion='entropy', random_state=42)
clf_id3.fit(X_train, y_train)
pred_id3 = clf_id3.predict(X_test)
acc_id3 = metrics.accuracy_score(y_test, pred_id3)
f1_id3 = metrics.f1_score(y_test, pred_id3, average='weighted', zero_division=0)

# 2. Cây CART
clf_cart = tree.DecisionTreeClassifier(criterion='gini', random_state=42)
clf_cart.fit(X_train, y_train)
pred_cart = clf_cart.predict(X_test)
acc_cart = metrics.accuracy_score(y_test, pred_cart)
f1_cart = metrics.f1_score(y_test, pred_cart, average='weighted', zero_division=0)

# 3. Naïve Bayes
clf_gnb = GaussianNB()
clf_gnb.fit(X_train, y_train)
pred_gnb = clf_gnb.predict(X_test)
acc_gnb = metrics.accuracy_score(y_test, pred_gnb)
f1_gnb = metrics.f1_score(y_test, pred_gnb, average='weighted', zero_division=0)

# 4. Bảng tổng hợp so sánh
df_compare = pd.DataFrame({
    'Model': ['Decision Tree (ID3)', 'Decision Tree (CART)', 'Gaussian Naïve Bayes'],
    'Accuracy': [acc_id3, acc_cart, acc_gnb],
    'F1-Score (Weighted)': [f1_id3, f1_cart, f1_gnb]
})
print("BẢNG SO SÁNH KẾT QUẢ TRÊN DỮ LIỆU RED WINE QUALITY:")
display(df_compare)

# 5. Biểu đồ so sánh
plt.figure(figsize=(10, 5))
bars = plt.bar(df_compare['Model'], df_compare['Accuracy'], color=['teal', 'coral', 'mediumpurple'], width=0.45)
plt.ylim(0.4, 0.7)
plt.ylabel('Độ chính xác (Accuracy)')
plt.title('So sánh độ chính xác giữa ID3, CART và Naïve Bayes trên Red Wine Quality')
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.005, f"{yval:.4f}", ha='center', va='bottom', fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 6. Nhận xét
print("=" * 60)
print("NHẬN XÉT:")
print("1. Cây quyết định (ID3 và CART) đạt độ chính xác từ 56% - 58%, vượt trội so với Naïve Bayes (~54%).")
print("2. Đối với bài toán phân nhiều lớp chất lượng rượu (từ 3 đến 8), cây quyết định phân tách ranh giới các lớp phi tuyến hiệu quả hơn.")
print("3. Naïve Bayes gặp khó khăn do phân phối các mức điểm 3, 4, 8 rất ít (mất cân bằng lớp nghiêm trọng).")
```
