```python
# ========================================================
# BÀI 6 (TRANG 27) - CÂU 6
# Đề bài: Sử dụng câu lệnh chia dữ liệu đầu vào thành hai phần
# huấn luyện 70% và kiểm thử 30% trên tập dữ liệu Mushroom.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

from sklearn.model_selection import train_test_split
import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


mush_path = find_file("mushrooms.csv")
data = pd.read_csv(mush_path)

features = data.drop('class', axis=1)
labels = data['class']

features_onehot = pd.get_dummies(features)

# 2. Chia tập dữ liệu 70% huấn luyện và 30% kiểm thử
X_train, X_test, y_train, y_test = train_test_split(features_onehot, labels, test_size=0.3, random_state=42)

print(f"Tổng số mẫu trong dữ liệu nấm: {len(data)}")
print(f"Số lượng mẫu huấn luyện (X_train): {X_train.shape[0]} ({len(X_train)/len(data)*100:.1f}%)")
print(f"Số lượng mẫu kiểm thử (X_test): {X_test.shape[0]} ({len(X_test)/len(data)*100:.1f}%)")
print("-" * 60)
print(f"Kích thước X_train: {X_train.shape}, y_train: {y_train.shape}")
print(f"Kích thước X_test: {X_test.shape}, y_test: {y_test.shape}")
```
