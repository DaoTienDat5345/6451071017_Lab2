```python
# ========================================================
# BÀI 6 (TRANG 27) - CÂU 5
# Đề bài: Chuyển đổi các cột không phải dạng số về dạng one-hot vector
# để phù hợp với đầu vào của thư viện (áp dụng trên dữ liệu Mushroom).
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


mush_path = find_file("mushrooms.csv")
data = pd.read_csv(mush_path)
features = data.drop('class', axis=1)

print("Kiểu dữ liệu của các cột trong tập features:")
print(features.dtypes.value_counts())
print("-" * 60)

# 2. Chuyển đổi toàn bộ các cột phân loại (categorical) sang one-hot vector
features_onehot = pd.get_dummies(features)

print(f"Số lượng thuộc tính trước khi chuyển đổi: {features.shape[1]}")
print(f"Số lượng thuộc tính sau khi mã hóa one-hot: {features_onehot.shape[1]}")
print(f"Kích thước ma trận features_onehot: {features_onehot.shape}")
print("-" * 60)
print("5 dòng đầu sau khi chuyển đổi sang one-hot vector:")
display(features_onehot.head())
```
