```python
# ========================================================
# BÀI 6 (TRANG 27) - CÂU 7
# Đề bài: Xây dựng cây ID3 (criterion='entropy') trên dữ liệu Mushroom,
# kiểm thử kết quả bằng ma trận nhầm lẫn và biểu diễn cây.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn import tree, metrics
import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


data = pd.read_csv(find_file("mushrooms.csv"))
features = data.drop('class', axis=1)
labels = data['class']
features_onehot = pd.get_dummies(features)

X_train, X_test, y_train, y_test = train_test_split(features_onehot, labels, test_size=0.3, random_state=42)

# 2. Xây dựng cây ID3
clf = tree.DecisionTreeClassifier(criterion='entropy', random_state=42)
clf.fit(X_train, y_train)

# 3. Dự đoán trên tập kiểm thử
tree_pred = clf.predict(X_test)

# 4. Đánh giá độ chính xác
tree_score = metrics.accuracy_score(y_test, tree_pred)
print("Accuracy (ID3):", tree_score)
print("Report:\n", metrics.classification_report(y_test, tree_pred))

# 5. Ma trận nhầm lẫn Heatmap
tree_cm = metrics.confusion_matrix(y_test, tree_pred)
plt.figure(figsize=(7, 6))
sns.heatmap(tree_cm, annot=True, fmt="d", cmap='Blues_r', square=True,
            xticklabels=clf.classes_, yticklabels=clf.classes_)
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
plt.title(f"ID3 Decision Tree (Mushroom) - Accuracy: {tree_score:.4f}", size=14)
plt.show()

# 6. Biểu diễn cây ID3
fig, ax = plt.subplots(figsize=(20, 8))
tree.plot_tree(clf, max_depth=3, feature_names=list(features_onehot.columns), class_names=list(clf.classes_), filled=True, fontsize=10)
plt.title("Cấu trúc cây quyết định ID3 trên tập dữ liệu nấm (Mushroom)", fontsize=15)
plt.savefig('mushroom_decision_tree_ID3.png', dpi=100)
plt.show()
```
