```python
# ========================================================
# BÀI 6 (TRANG 27) - CÂU 10
# Đề bài: So sánh kết quả của các mô hình trên (ID3, CART, Naïve Bayes)
# trên tập dữ liệu Mushroom.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

%matplotlib inline
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn import tree, metrics
from sklearn.naive_bayes import GaussianNB
import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


data = pd.read_csv(find_file("mushrooms.csv"))
features = data.drop('class', axis=1)
labels = data['class']
features_onehot = pd.get_dummies(features)

X_train, X_test, y_train, y_test = train_test_split(features_onehot, labels, test_size=0.3, random_state=42)

# 1. Cây ID3
clf_id3 = tree.DecisionTreeClassifier(criterion='entropy', random_state=42)
clf_id3.fit(X_train, y_train)
pred_id3 = clf_id3.predict(X_test)
acc_id3 = metrics.accuracy_score(y_test, pred_id3)
f1_id3 = metrics.f1_score(y_test, pred_id3, pos_label='p')

# 2. Cây CART
clf_cart = tree.DecisionTreeClassifier(criterion='gini', random_state=42)
clf_cart.fit(X_train, y_train)
pred_cart = clf_cart.predict(X_test)
acc_cart = metrics.accuracy_score(y_test, pred_cart)
f1_cart = metrics.f1_score(y_test, pred_cart, pos_label='p')

# 3. Naïve Bayes
clf_gnb = GaussianNB()
clf_gnb.fit(X_train, y_train)
pred_gnb = clf_gnb.predict(X_test)
acc_gnb = metrics.accuracy_score(y_test, pred_gnb)
f1_gnb = metrics.f1_score(y_test, pred_gnb, pos_label='p')

# 4. Bảng tổng hợp so sánh
df_compare = pd.DataFrame({
    'Model': ['Decision Tree (ID3)', 'Decision Tree (CART)', 'Gaussian Naïve Bayes'],
    'Accuracy': [acc_id3, acc_cart, acc_gnb],
    'F1-Score (Class Poisonous)': [f1_id3, f1_cart, f1_gnb]
})
print("BẢNG TỔNG HỢP SO SÁNH HIỆU NĂNG TRÊN TẬP MUSHROOM:")
display(df_compare)

# 5. Biểu đồ so sánh
plt.figure(figsize=(10, 5))
bars = plt.bar(df_compare['Model'], df_compare['Accuracy'], color=['dodgerblue', 'forestgreen', 'darkorange'], width=0.45)
plt.ylim(0.9, 1.05)
plt.ylabel('Độ chính xác (Accuracy)')
plt.title('So sánh độ chính xác giữa ID3, CART và Naïve Bayes trên tập Mushroom')
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval + 0.005, f"{yval:.4f}", ha='center', va='bottom', fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# 6. Nhận xét
print("=" * 60)
print("NHẬN XÉT VÀ KẾT LUẬN:")
print("1. Cả hai mô hình Cây quyết định (ID3 và CART) đều đạt độ chính xác tuyệt đối 100% (Accuracy = 1.0) trên tập kiểm thử.")
print("2. Dữ liệu Mushroom có các thuộc tính đặc trưng mang tính phân biệt rất cao (như mùi hương odor, màu vết bầm, cấu trúc cuống), giúp cây chỉ cần một vài phép tách là phân loại hoàn hảo.")
print("3. Naïve Bayes cũng đạt kết quả rất tốt (~96%), tuy nhiên thấp hơn cây quyết định do giả định độc lập không hoàn toàn thỏa mãn giữa các đặc tính hình thái của nấm.")
```
