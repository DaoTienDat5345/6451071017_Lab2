```python
# ========================================================
# BÀI 6 (TRANG 27) - CÂU 4
# Đề bài: Cho dữ liệu Mushroom (mushrooms.csv), chứa thông tin đặc điểm
# của nhiều loại nấm cùng với phân loại 'class' là nấm độc (p) hay ăn được (e).
# Tách các cột dữ liệu thành hai phần: features và labels ('class').
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


# 2. Đọc file mushrooms.csv
mush_path = find_file("mushrooms.csv")
data = pd.read_csv(mush_path)

# 3. Tách thuộc tính đặc trưng và nhãn quyết định
features = data.drop('class', axis=1)
labels = data['class']

print(f"Kích thước tập dữ liệu gốc: {data.shape}")
print(f"Kích thước tập features: {features.shape}")
print(f"Kích thước thuộc tính quyết định labels (class): {labels.shape}")
print("-" * 60)
print("Danh sách các thuộc tính:")
print(list(features.columns))
print("-" * 60)
print("Phân phối của nhãn nấm (p: nấm độc, e: nấm ăn được):")
print(labels.value_counts())
print("-" * 60)
display(features.head())
```
