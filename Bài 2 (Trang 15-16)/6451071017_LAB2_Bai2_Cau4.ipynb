```python
# ========================================================
# BÀI 2 (TRANG 15-16) - CÂU 4
# Đề bài: Tách các cột dữ liệu thành hai phần, một phần chứa các thuộc tính
# bình thường, một phần chứa riêng thuộc tính quyết định.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


# Tiền xử lý dữ liệu
train_data = pd.read_csv(find_file("adult.data.csv"), skipinitialspace=True, na_values='?').dropna()
test_data = pd.read_csv(find_file("adult.test.csv"), skipinitialspace=True, na_values='?').dropna()
if 'fnlwgt' in train_data.columns:
    train_data = train_data.drop(columns=['fnlwgt'])
data = pd.concat([train_data, test_data], ignore_index=True)

# 2. Tách dữ liệu vào biến features và labels
features = data.drop('income', axis=1)
labels = data['income']

# 3. Hiển thị thông tin
print(f"Kích thước của tập thuộc tính features: {features.shape}")
print(f"Kích thước của thuộc tính quyết định labels: {labels.shape}")
print("-" * 60)
print("Danh sách các thuộc tính trong features:")
print(list(features.columns))
print("-" * 60)
print("Phân phối các nhãn trong labels (income):")
print(labels.value_counts())
print("-" * 60)
display(features.head())
```
