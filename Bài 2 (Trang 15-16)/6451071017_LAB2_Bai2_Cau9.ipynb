```python
# ========================================================
# BÀI 2 (TRANG 15-16) - CÂU 9
# Đề bài: Xây dựng mô hình phân lớp bằng thuật toán Naïve Bayes
# và kiểm tra kết quả đạt được.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.naive_bayes import GaussianNB
from sklearn import metrics
import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


# Chuẩn bị dữ liệu
train_data = pd.read_csv(find_file("adult.data.csv"), skipinitialspace=True, na_values='?').dropna()
test_data = pd.read_csv(find_file("adult.test.csv"), skipinitialspace=True, na_values='?').dropna()
num_train = len(train_data)

if 'fnlwgt' in train_data.columns:
    train_data = train_data.drop(columns=['fnlwgt'])
data = pd.concat([train_data, test_data], ignore_index=True)

features = data.drop('income', axis=1)
labels = data['income']
cat_cols = features.select_dtypes(exclude=['int64', 'float64']).columns
features_onehot = pd.get_dummies(features, columns=cat_cols)

X_train = features_onehot[:num_train]
X_test = features_onehot[num_train:]
y_train = labels[:num_train]
y_test = labels[num_train:]

# 2. Xây dựng mô hình Gaussian Naïve Bayes
gnb = GaussianNB()
bayes_pred = gnb.fit(X_train, y_train).predict(X_test)

# 3. Đánh giá mô hình
bayes_score = metrics.accuracy_score(y_test, bayes_pred)
print("Accuracy (Naïve Bayes):", bayes_score)
print("Report:\n", metrics.classification_report(y_test, bayes_pred))

# 4. Biểu diễn ma trận nhầm lẫn bằng đồ thị heatmap
bayes_cm = metrics.confusion_matrix(y_test, bayes_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(bayes_cm, annot=True, fmt="d", linewidths=.5, square=True, cmap='Blues_r',
            xticklabels=gnb.classes_, yticklabels=gnb.classes_)
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
title = 'Naïve Bayes Accuracy Score: 0.0000'.format(bayes_score)
plt.title(title, size=15)
plt.show()
```
