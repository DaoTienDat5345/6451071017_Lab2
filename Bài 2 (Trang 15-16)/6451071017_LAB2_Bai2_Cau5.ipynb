```python
# ========================================================
# BÀI 2 (TRANG 15-16) - CÂU 5
# Đề bài: Chuyển đổi các cột không phải dạng số về dạng one-hot vector
# để phù hợp với đầu vào của thư viện.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


train_data = pd.read_csv(find_file("adult.data.csv"), skipinitialspace=True, na_values='?').dropna()
test_data = pd.read_csv(find_file("adult.test.csv"), skipinitialspace=True, na_values='?').dropna()
if 'fnlwgt' in train_data.columns:
    train_data = train_data.drop(columns=['fnlwgt'])
data = pd.concat([train_data, test_data], ignore_index=True)
features = data.drop('income', axis=1)

# 2. Xác định các thuộc tính không phải dạng số trong biến features
categorical_cols = features.select_dtypes(exclude=['int64', 'float64']).columns
print("Các thuộc tính không phải dạng số cần chuyển đổi:")
print(categorical_cols)
print("-" * 60)

# 3. Kết hợp pd.get_dummies để chuyển đổi sang dạng one-hot vector
features_onehot = pd.get_dummies(features, columns=categorical_cols)

print(f"Kích thước features trước khi mã hóa: {features.shape}")
print(f"Kích thước features sau khi mã hóa one-hot: {features_onehot.shape}")
print("-" * 60)
print("5 dòng đầu sau khi chuyển đổi dạng one-hot:")
display(features_onehot.head())
```
