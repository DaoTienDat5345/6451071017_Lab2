```python
# ========================================================
# BÀI 2 (TRANG 15-16) - CÂU 8
# Đề bài: Lặp lại yêu cầu 7 đối với cây CART (criterion='gini').
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import tree, metrics
import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


# Chuẩn bị dữ liệu
train_data = pd.read_csv(find_file("adult.data.csv"), skipinitialspace=True, na_values='?').dropna()
test_data = pd.read_csv(find_file("adult.test.csv"), skipinitialspace=True, na_values='?').dropna()
num_train = len(train_data)

if 'fnlwgt' in train_data.columns:
    train_data = train_data.drop(columns=['fnlwgt'])
data = pd.concat([train_data, test_data], ignore_index=True)

features = data.drop('income', axis=1)
labels = data['income']
cat_cols = features.select_dtypes(exclude=['int64', 'float64']).columns
features_onehot = pd.get_dummies(features, columns=cat_cols)

X_train = features_onehot[:num_train]
X_test = features_onehot[num_train:]
y_train = labels[:num_train]
y_test = labels[num_train:]

# 2. Xây dựng cây CART với criterion='gini'
clf_cart = tree.DecisionTreeClassifier(criterion='gini', random_state=42)
clf_cart = clf_cart.fit(X_train, y_train)

# 3. Dự đoán trên tập kiểm thử
cart_pred = clf_cart.predict(X_test)

# 4. Đánh giá độ chính xác và in báo cáo phân lớp
cart_score = metrics.accuracy_score(y_test, cart_pred)
print("Accuracy (CART):", cart_score)
print("Report:\n", metrics.classification_report(y_test, cart_pred))

# 5. Tính ma trận nhầm lẫn và biểu diễn lên heatmap
cart_cm = metrics.confusion_matrix(y_test, cart_pred)
plt.figure(figsize=(10, 8))
sns.heatmap(cart_cm, annot=True, fmt="d", linewidths=.5, square=True, cmap='Blues_r',
            xticklabels=clf_cart.classes_, yticklabels=clf_cart.classes_)
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
title = 'CART Decision Tree Accuracy Score: 0.0000'.format(cart_score)
plt.title(title, size=15)
plt.show()

# 6. Biểu diễn cây CART
fig, ax = plt.subplots(figsize=(24, 12))
tree.plot_tree(clf_cart, max_depth=3, feature_names=list(X_train.columns), class_names=list(clf_cart.classes_), filled=True, fontsize=10)
plt.title("Biểu diễn cây quyết định CART (tối đa độ sâu 3)", fontsize=16)
plt.savefig('decision_tree_CART.png', dpi=100)
plt.show()
print("Đã lưu hình ảnh cây CART vào file decision_tree_CART.png")
```
