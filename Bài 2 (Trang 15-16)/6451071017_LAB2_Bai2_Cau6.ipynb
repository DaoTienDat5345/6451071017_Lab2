```python
# ========================================================
# BÀI 2 (TRANG 15-16) - CÂU 6
# Đề bài: Tách các dòng dữ liệu ra thành hai phần huấn luyện và kiểm thử
# như ban đầu.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


train_data = pd.read_csv(find_file("adult.data.csv"), skipinitialspace=True, na_values='?').dropna()
test_data = pd.read_csv(find_file("adult.test.csv"), skipinitialspace=True, na_values='?').dropna()
num_train = len(train_data) # 30162
num_test = len(test_data)   # 15060

if 'fnlwgt' in train_data.columns:
    train_data = train_data.drop(columns=['fnlwgt'])
data = pd.concat([train_data, test_data], ignore_index=True)

features = data.drop('income', axis=1)
labels = data['income']

categorical_cols = features.select_dtypes(exclude=['int64', 'float64']).columns
features_onehot = pd.get_dummies(features, columns=categorical_cols)

# 2. Tách lại dữ liệu huấn luyện và kiểm thử như thông tin thu được ban đầu
X_train = features_onehot[:num_train]
X_test = features_onehot[num_train:]
y_train = labels[:num_train]
y_test = labels[num_train:]

print(f"Số lượng mẫu tập huấn luyện (X_train): {X_train.shape}")
print(f"Số lượng mẫu tập kiểm thử (X_test): {X_test.shape}")
print(f"Số lượng nhãn tập huấn luyện (y_train): {y_train.shape}")
print(f"Số lượng nhãn tập kiểm thử (y_test): {y_test.shape}")
```
