```python
# ========================================================
# BÀI 2 (TRANG 15-16) - CÂU 1
# Đề bài: Nhập dữ liệu đầu vào và cho biết số lượng dữ liệu huấn luyện
# (tương ứng với file adult.data.csv) và số lượng dữ liệu kiểm thử
# (tương ứng với file adult.test.csv).
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name

# 2. Đọc dữ liệu huấn luyện và kiểm thử
train_path = find_file("adult.data.csv")
test_path = find_file("adult.test.csv")

train_data = pd.read_csv(train_path, skipinitialspace=True)
test_data = pd.read_csv(test_path, skipinitialspace=True)

# 3. Hiển thị số lượng dữ liệu
print(f"Number of training data: {len(train_data)}")
print(f"Number of test data: {len(test_data)}")
print("-" * 60)

print("5 dòng đầu của dữ liệu huấn luyện (adult.data.csv):")
display(train_data.head())

print("5 dòng đầu của dữ liệu kiểm thử (adult.test.csv):")
display(test_data.head())
```
