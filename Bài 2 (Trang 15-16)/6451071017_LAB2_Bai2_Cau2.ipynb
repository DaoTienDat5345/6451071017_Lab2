```python
# ========================================================
# BÀI 2 (TRANG 15-16) - CÂU 2
# Đề bài: Tiến hành tiền xử lý dữ liệu:
# - Xóa những dòng có chứa dữ liệu trống, biết rằng dữ liệu trống được ký hiệu bằng dấu '?'.
# - Xóa cột final weight 'fnlwgt' trong dữ liệu huấn luyện vì cột này không có trong dữ liệu kiểm thử.
# - Nối dữ liệu huấn luyện và kiểm thử lại với nhau để phục vụ các bước tiếp theo.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name

train_path = find_file("adult.data.csv")
test_path = find_file("adult.test.csv")

# 2. Đọc dữ liệu và xác định ký hiệu '?' là giá trị rỗng (NaN)
train_data = pd.read_csv(train_path, skipinitialspace=True, na_values='?')
test_data = pd.read_csv(test_path, skipinitialspace=True, na_values='?')

print("--- TRƯỚC KHI TIỀN XỬ LÝ ---")
print(f"Số dòng train ban đầu: {len(train_data)}")
print(f"Số dòng test ban đầu: {len(test_data)}")

# 3. Xóa các dòng chứa giá trị rỗng '?'
train_data = train_data.dropna()
test_data = test_data.dropna()

print("\n--- SAU KHI XÓA CÁC DÒNG RỖNG ('?') ---")
print(f"Number of training data: {len(train_data)}")
print(f"Number of test data: {len(test_data)}")

# 4. Xóa cột 'fnlwgt' trong dữ liệu huấn luyện
if 'fnlwgt' in train_data.columns:
    train_data = train_data.drop(columns=['fnlwgt'])
    print("\nĐã xóa cột 'fnlwgt' trong tập huấn luyện thành công.")

# 5. Nối dữ liệu huấn luyện và kiểm thử lại với nhau vào biến data
data = pd.concat([train_data, test_data], ignore_index=True)
print(f"\nKích thước tập dữ liệu hợp nhất (data): {data.shape}")
print("-" * 60)
print("Thông tin khối dữ liệu sau khi tiền xử lý:")
data.info()
print("-" * 60)
display(data.head())
```
