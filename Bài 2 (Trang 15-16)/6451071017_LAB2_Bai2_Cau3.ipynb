```python
# ========================================================
# BÀI 2 (TRANG 15-16) - CÂU 3
# Đề bài: Khảo sát độ tương đồng giữa các cột với nhau bằng công thức Pearson.
# Những cột nào có giá trị tương đồng cao thì hãy loại bỏ.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sns
import os
import pandas as pd
import numpy as np

def find_file(name):
    paths = [
        name,
        os.path.join(r"D:\khai pha du lieu\Lab2\Dataset-20261103T094921Z-1-001\Dataset", name),
        os.path.join("..", "..", "..", "Dataset-20261103T094921Z-1-001", "Dataset", name),
        os.path.join("Dataset", name)
    ]
    for p in paths:
        if os.path.exists(p):
            return p
    return name


# 2. Tiền xử lý dữ liệu để có biến data
train_data = pd.read_csv(find_file("adult.data.csv"), skipinitialspace=True, na_values='?').dropna()
test_data = pd.read_csv(find_file("adult.test.csv"), skipinitialspace=True, na_values='?').dropna()
if 'fnlwgt' in train_data.columns:
    train_data = train_data.drop(columns=['fnlwgt'])
data = pd.concat([train_data, test_data], ignore_index=True)

# 3. Khảo sát độ tương đồng Pearson giữa các cột số
numeric_data = data.select_dtypes(include=['int64', 'float64'])
corr_matrix = numeric_data.corr(method='pearson')

print("Ma trận tương quan Pearson giữa các thuộc tính số:")
display(corr_matrix)

# 4. Trực quan hóa kết quả bằng đồ thị heatmap
plt.figure(figsize=(16, 9))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title("Heatmap độ tương đồng Pearson giữa các thuộc tính", fontsize=16)
plt.show()

# 5. Nhận xét
print("=" * 60)
print("NHẬN XÉT:")
print("- Trong đồ thị heatmap trên, màu càng đậm chứng tỏ mức độ tương đồng càng thấp.")
print("- Nhận thấy không có cặp thuộc tính nào có độ tương quan quá cao (tất cả đều < 0.5).")
print("- Do đó, không cần thiết phải loại bỏ thuộc tính nào.")
```
