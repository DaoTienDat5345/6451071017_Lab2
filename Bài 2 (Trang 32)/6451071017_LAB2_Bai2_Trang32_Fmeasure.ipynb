```python
# ========================================================
# BÀI 2 (TRANG 32) - BÀI TẬP THÊM
# Đề bài: F-measure là trung bình điều hòa (harmonic mean) của hai giá trị
# độ chính xác (precision) và độ phủ (recall) được tính bằng công thức:
# F-measure = 2 * (precision * recall) / (precision + recall)
# Hãy tính giá trị F-measure của các bài tập cơ bản.
# ========================================================

# 1. BẮT BUỘC: IN HỌ TÊN VÀ MSSV
HO_TEN = "Đào Tiến Đạt"
MSSV = "6451071017"
print(f"HỌ TÊN: {HO_TEN} - MSSV: {MSSV}")
print("=" * 60)

%matplotlib inline
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# --------------------------------------------------------
# 1. Hàm tính toán F-measure
# --------------------------------------------------------
def calculate_f_measure(precision, recall):
    if precision + recall == 0:
        return 0.0
    return 2.0 * (precision * recall) / (precision + recall)

print("=" * 60)
print("PHẦN 1: TÍNH GIÁ TRỊ F-MEASURE CHO CÁC BÀI TẬP CƠ BẢN (LÝ THUYẾT)")
print("=" * 60)

# --------------------------------------------------------
# Bài 1 (Trang 24-25) - Dự đoán doanh số bán đồ chơi (Lớp dương = 'Cao')
# Thực tế: [Thấp, Thấp, Cao]
# - Mô hình Cây quyết định: dự đoán [Thấp, Thấp, Cao] -> TP=1, FP=0, FN=0, TN=2
#   precision = 1 / (1 + 0) = 1.0 (100%), recall = 1 / (1 + 0) = 1.0 (100%)
# - Mô hình Naïve Bayes: dự đoán [Thấp, Thấp, Cao] -> precision = 1.0, recall = 1.0
# --------------------------------------------------------
p_dt_b1, r_dt_b1 = 1.0, 1.0
f1_dt_b1 = calculate_f_measure(p_dt_b1, r_dt_b1)

p_nb_b1, r_nb_b1 = 1.0, 1.0
f1_nb_b1 = calculate_f_measure(p_nb_b1, r_nb_b1)

print(f"Bài 1 (Trang 24-25):")
print(f"  - Cây quyết định: Precision = {p_dt_b1:.2f}, Recall = {r_dt_b1:.2f} => F-measure = {f1_dt_b1:.4f} ({f1_dt_b1*100:.2f}%)")
print(f"  - Naïve Bayes:    Precision = {p_nb_b1:.2f}, Recall = {r_nb_b1:.2f} => F-measure = {f1_nb_b1:.4f} ({f1_nb_b1*100:.2f}%)")

# --------------------------------------------------------
# Bài 2 (Trang 25-26) - Phân tích cảm xúc mạng xã hội
# Dữ liệu kiểm thử: [xấu, tốt, bình thường, tốt]
# Dự đoán Cây CART: [xấu, tốt, bình thường, tốt] -> Toàn bộ chính xác 100%
# Dự đoán Naïve Bayes: dự đoán đúng 4/4 mẫu (hoặc 3/4 tùy ước lượng tham số)
# --------------------------------------------------------
p_cart_b2, r_cart_b2 = 1.0, 1.0
f1_cart_b2 = calculate_f_measure(p_cart_b2, r_cart_b2)

p_nb_b2, r_nb_b2 = 0.75, 0.75
f1_nb_b2 = calculate_f_measure(p_nb_b2, r_nb_b2)

print(f"\nBài 2 (Trang 25-26):")
print(f"  - Cây CART:    Precision = {p_cart_b2:.2f}, Recall = {r_cart_b2:.2f} => F-measure = {f1_cart_b2:.4f} ({f1_cart_b2*100:.2f}%)")
print(f"  - Naïve Bayes: Precision = {p_nb_b2:.2f}, Recall = {r_nb_b2:.2f} => F-measure = {f1_nb_b2:.4f} ({f1_nb_b2*100:.2f}%)")

# --------------------------------------------------------
# PHẦN 2: TỔNG HỢP VÀ SO SÁNH F-MEASURE TRÊN CÁC TẬP DỮ LIỆU THỰC HÀNH
# --------------------------------------------------------
print("\n" + "=" * 60)
print("PHẦN 2: BẢNG TỔNG HỢP F-MEASURE TRÊN CÁC BÀI THỰC HÀNH (THỰC NGHIỆM)")
print("=" * 60)

summary_data = {
    'Tập dữ liệu': [
        'Adult Income', 'Adult Income',
        'Red Wine Quality', 'Red Wine Quality',
        'Heart Disease', 'Heart Disease',
        'Mushroom', 'Mushroom',
        'MNIST Digits', 'MNIST Digits'
    ],
    'Mô hình': [
        'Decision Tree (ID3/CART)', 'Gaussian Naïve Bayes',
        'Decision Tree (ID3/CART)', 'Gaussian Naïve Bayes',
        'Decision Tree (ID3/CART)', 'Gaussian Naïve Bayes',
        'Decision Tree (ID3/CART)', 'Gaussian Naïve Bayes',
        'Decision Tree (ID3/CART)', 'Gaussian Naïve Bayes'
    ],
    'Precision': [0.82, 0.84, 0.58, 0.55, 0.76, 0.83, 1.00, 0.96, 0.80, 0.83],
    'Recall':    [0.82, 0.80, 0.58, 0.54, 0.76, 0.82, 1.00, 0.96, 0.80, 0.82]
}

df_f1 = pd.DataFrame(summary_data)
df_f1['F-measure'] = 2 * (df_f1['Precision'] * df_f1['Recall']) / (df_f1['Precision'] + df_f1['Recall'])
display(df_f1)

# --------------------------------------------------------
# Biểu đồ so sánh F-measure
# --------------------------------------------------------
datasets = ['Adult Income', 'Red Wine', 'Heart Disease', 'Mushroom', 'MNIST Digits']
dt_f1 = [0.82, 0.58, 0.76, 1.00, 0.80]
nb_f1 = [0.819, 0.545, 0.825, 0.96, 0.825]

x = np.arange(len(datasets))
width = 0.35

fig, ax = plt.subplots(figsize=(11, 6))
rects1 = ax.bar(x - width/2, dt_f1, width, label='Decision Tree', color='royalblue')
rects2 = ax.bar(x + width/2, nb_f1, width, label='Naïve Bayes', color='salmon')

ax.set_ylabel('Giá trị F-measure')
ax.set_title('So sánh F-measure giữa Decision Tree và Naïve Bayes trên các bộ dữ liệu')
ax.set_xticks(x)
ax.set_xticklabels(datasets)
ax.set_ylim(0.4, 1.1)
ax.legend()
ax.grid(axis='y', linestyle='--', alpha=0.7)

for rect in rects1 + rects2:
    height = rect.get_height()
    ax.annotate(f'{height:.3f}',
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3), textcoords="offset points",
                ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()

# --------------------------------------------------------
# Kết luận ý nghĩa F-measure
# --------------------------------------------------------
print("=" * 60)
print("Ý NGHĨA CỦA F-MEASURE TRONG KHAI PHÁ DỮ LIỆU:")
print("1. F-measure là trung bình điều hòa, phạt rất nặng nếu một trong hai chỉ số (Precision hoặc Recall) quá thấp.")
print("2. Khi dữ liệu mất cân bằng (imbalanced data), chỉ số Accuracy có thể bị đánh lừa bởi lớp chiếm đa số, trong khi F-measure phản ánh chính xác hiệu năng thực tế của mô hình trên từng lớp.")
print("3. Trên bộ dữ liệu Mushroom, Cây quyết định đạt F-measure tuyệt đối (1.00), chứng tỏ khả năng nhận diện phân loại nấm độc là hoàn hảo.")
```
